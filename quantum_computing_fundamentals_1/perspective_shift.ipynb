## Think of a decision you made where you couldn't know the outcome until you committed to it. How is that like measurement collapsing a qubit?

### A decision is choosing what I want to do as an occupation. There are thousands of different jobs in the world, but you won't know what you will do unless you commit to one thing. It is like measuring a qubit. Before deciding many futures are possible. Once I commit, those possibilities narrow down to one path, similar to how a qubit causes it to collapse from multiple possible states into one result.

## Have you ever been two completely different things at once -- depending on who was watching? What does superposition say about that?

### Sometimes I seem like two different people depending on whos around me. For instance, I might be quiet and focused in class but outgoing with my close friends. Superposition reminds me of this because, it means a qubit can exist in multiple states at the same time until it is measured. It's like having different possible outcomes that all coexist before one is observed. While people can't be actually in superposition, the idea of seeming like being different people depending on the situation is a good analogy.
